```python
# Problem 1: Zipper file()

def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            zipped_string = strings[i] + strings[i + 1]
            file.write(zipped_string + "\n")

    print("File created successfully!")

    with open(filename, "r") as file:
        print("Contents of the file:")
        print(file.read())


# Input filename
filename = "zipped_strings.txt"

# List of strings
strings = ["Hello", "World", "Python", "Programming"]

# Calling the function
zipper_file(filename, strings)
```

```python
# Problem 2: Reformat date()

def reformat_date(dates):

    # Dictionary containing month names
    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    # Convert each date into the required format
    for date in dates:
        month = date[0:2]
        day = date[3:5]
        year = date[6:10]

        month_name = months[month]

        result = "Office Hours on " + month_name + " " + day + ", " + year

        print(result)


# List of dates in MM/DD/YYYY format
dates = ["01/15/2025", "03/09/2025", "11/25/2025", "12/05/2025"]

# Calling the function
reformat_date(dates)
```
