# Practical 10 - Visualization

## Aim
To visualize the Iris dataset using Seaborn library and perform analysis on Scatter Plots, Density Plots, Violin Plots, Heatmaps, Joint Plots, and Pair Plots.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("Libraries imported successfully!")
```

```python
iris = sns.load_dataset("iris")

print(iris.head())
```

```python
print("Dataset Shape:", iris.shape)

print("\nDataset Information:")
print(iris.info())

print("\nSpecies present in the dataset:")
print(iris["species"].unique())
```

## 1. Scatter Plot

```python
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")

plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

### Analysis

The scatter plot shows the relationship between sepal length and sepal width for different species of Iris flowers. Setosa flowers are generally distinguishable from the other two species. However, Versicolor and Virginica show some overlap. The plot indicates that sepal length and sepal width alone are not sufficient to clearly separate all three species.

## 2. Density Plot

```python
sns.kdeplot(data=iris, x="petal_length", hue="species", fill=True)

plt.title("Density Plot of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Density")
plt.show()
```

### Analysis

The density plot shows the distribution of petal length for different Iris species. Setosa has a shorter petal length and is clearly separated from the other species. Versicolor has medium petal lengths, while Virginica generally has longer petal lengths. The distributions of Versicolor and Virginica overlap slightly, but petal length helps distinguish the species.

## 3. Violin Plot

```python
sns.violinplot(data=iris, x="species", y="petal_length")

plt.title("Violin Plot of Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

### Analysis

The violin plot compares the distribution of petal lengths among the three Iris species. Setosa has the smallest petal lengths, while Virginica has the largest. Versicolor lies between the two. The width of each violin represents the density of observations at different petal lengths. The plot shows that petal length is an important feature for distinguishing the Iris species.

## 4. Heatmap

```python
correlation = iris.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

### Analysis

The heatmap represents the correlation between the numerical features of the Iris dataset. Petal length and petal width show a strong positive correlation, meaning that flowers with longer petals generally have wider petals. Sepal length also shows a strong positive correlation with petal length and petal width. Sepal width has a comparatively weaker correlation with the other features. The heatmap helps identify relationships between variables and shows that petal measurements are strongly related.

## 5. Joint Plot

```python
sns.jointplot(data=iris, x="petal_length", y="petal_width", hue="species")

plt.show()
```

### Analysis

The joint plot shows the relationship between petal length and petal width along with their individual distributions. A strong positive relationship can be observed between petal length and petal width. Setosa flowers form a separate cluster with smaller petal measurements. Versicolor and Virginica have larger petal measurements and show some overlap. This plot helps understand both the relationship and distribution of the two variables.

## 6. Pair Plot

```python
sns.pairplot(iris, hue="species")

plt.show()
```

### Analysis

The pair plot displays relationships between all numerical features of the Iris dataset. Setosa is clearly separated from the other species in plots involving petal length and petal width. Versicolor and Virginica show some overlap but can be distinguished to a certain extent. The diagonal plots show the distribution of individual features. Overall, petal length and petal width are the most useful features for distinguishing the three species.

## Conclusion

In this practical, the Iris dataset was visualized using the Seaborn library. Scatter plots, density plots, violin plots, heatmaps, joint plots, and pair plots were created and analyzed. These visualizations helped understand the distribution of data, relationships between different features, and differences between Iris species. The heatmap showed correlations between numerical variables, while the pair plot and other visualizations demonstrated that petal length and petal width are useful features for distinguishing the species.
