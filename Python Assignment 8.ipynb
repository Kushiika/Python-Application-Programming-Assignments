```python
## Problem 4: Data Loading and Exploration
```

```python
customer_data = {
    "Customer_ID": [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    
    "Name": ["Aarav", "Diya", "Rohan", "Ananya",
             "Kabir", "Meera", "Arjun", "Isha",
             "Vivaan", "Sara"],
    
    "Age": [25, 34, 29, 42, np.nan,
            31, 45, 28, 36, 23],
    
    "Gender": ["Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female",
               "Male", "Female"],
    
    "Purchase_Amount": [2500, 4000, 1800, 5200, 3000,
                        4500, 6000, np.nan, 3500, 2200],
    
    "City": ["Delhi", "Mumbai", "Chandigarh", "Delhi",
             "Jaipur", "Mumbai", "Delhi", "Jaipur",
             "Chandigarh", "Mumbai"]
}

customers = pd.DataFrame(customer_data)

print("Customer Dataset:")
display(customers)
```

```python
print("First five rows:")
display(customers.head())
```

```python
print("Summary Statistics:")
display(customers.describe())
```

```python
print("Missing values in each column:")
print(customers.isnull().sum())
```

```python
customers["Age"] = customers["Age"].fillna(
    customers["Age"].median()
)

customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(
    customers["Purchase_Amount"].median()
)

print("Dataset after handling missing values:")
display(customers)
```

```python
print("Missing values after handling:")
print(customers.isnull().sum())
```

```python
## Problem 5: Data Filtering and Selection
```

```python
older_customers = customers[customers["Age"] > 30]

print("Customers whose age is greater than 30:")
display(older_customers)
```

```python
male_total = customers.loc[
    customers["Gender"] == "Male", "Purchase_Amount"
].sum()

female_total = customers.loc[
    customers["Gender"] == "Female", "Purchase_Amount"
].sum()

print("Total purchase amount for male customers:", male_total)
print("Total purchase amount for female customers:", female_total)
```

```python
selected_columns = customers[
    ["Customer_ID", "Name", "Purchase_Amount"]
]

print("Selected customer columns:")
display(selected_columns)
```

```python
## Problem 6: Data Transformation and Aggregation
```

```python
customers["Discounted_Price"] = (
    customers["Purchase_Amount"] * 0.90
)

print("Dataset with Discounted Price:")
display(customers)
```

```python
average_purchase = customers.groupby(
    "Gender"
)["Purchase_Amount"].mean()

print("Average Purchase Amount by Gender:")
print(average_purchase)
```

```python
sorted_customers = customers.sort_values(
    by="Purchase_Amount",
    ascending=False
)

print("Customers sorted by Purchase Amount (Descending):")
display(sorted_customers)
```
