```python
inventory={}

print("Empty Inventory:",inventory)
```

```python
product1_name="Mobile Phone"
product1_quantity=5
product1_price=20000
product1_releaseyear=2020

print("Product name:",product1_name)
print("Product quantity:",product1_quantity)
print("Product price:",product1_price)
print("Product release year:",product1_releaseyear)
```

```python
inventory["ProductNo1"] = {
    "Product Name": product1_name,
    "Product Quantity": product1_quantity,
    "Product Price": product1_price,
    "Product Release Year": product1_releaseyear}

print("Inventory after adding first product:")
print(inventory)
```

```python
product2_name = "Laptop"
product2_quantity = 3
product2_price = 55000
product2_releaseyear = 2021

print("Product Name:", product2_name)
print("Product Quantity:", product2_quantity)
print("Product Price:", product2_price)
print("Product Release Year:", product2_releaseyear)
```

```python
inventory["ProductNo2"] = {
    "Product Name": product2_name,
    "Product Quantity": product2_quantity,
    "Product Price": product2_price,
    "Product Release Year": product2_releaseyear}

print("Inventory after adding second product:")
print(inventory)
```

```python
print("Products present in inventory:")

for product_no, details in inventory.items():
    print("\n", product_no)

    for key, value in details.items():
        print(key, ":", value)
```

```python
check1 = "Product Release Year" in inventory["ProductNo1"]
check2 = "Product Release Year" in inventory["ProductNo2"]

print("ProductNo1_releaseyear in inventory:", check1)
print("ProductNo2_releaseyear in inventory:", check2)
```

```python
del inventory["ProductNo1"]["Product Release Year"]
del inventory["ProductNo2"]["Product Release Year"]

print("Inventory after deleting release year of both products:")
print(inventory)
```

```python
prices=(250,300,150,400,100,350,200)
print("Prices tuple:",prices)
```

```python
highest=max(prices)
lowest=min(prices)

print("Highest price:",highest)
print("Lowest price:",lowest)
```

```python
total_sum=sum(prices)

print("Total sum of all prices:",total_sum)
```

```python
sorted_list=sorted(prices)

print("Sorted list in ascending order:",sorted_list)
```

```python
try:
    prices[0] = 500
except TypeError:
    print("TypeError: Tuple elements cannot be modified.")
    print("Explanation: Tuples are immutable, so their elements cannot be changed.")
```
