```python
shopping_list=[]

print("Empty Shopping List:",shopping_list)
```

```python
shopping_list=["Watch","Laptop","Shoes","Pen","Clothes"]

print("Shopping List:",shopping_list)
```

```python
shopping_list.append("Football")

print("Shopping list after adding football:",shopping_list)
```

```python
last_item=shopping_list[-1]

print("Last item:",last_item)
```

```python
print("Entire shopping list:",shopping_list)
```

```python
print("Important items:")
print("Laptop:",shopping_list[1])
print("Shoes:",shopping_list[2])
```

```python
shopping_list[3]="notebook"

print("Shopping list after replacing pen with notebook:")
print(shopping_list)
```

```python
del shopping_list[0]

print("Final shopping list after deleting the least important item:")
print(shopping_list)
```
