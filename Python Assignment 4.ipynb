```python
players = {
    "Serena Williams": {"Sport": "Tennis", "Achievements": 23},
    "Lionel Messi": {"Sport": "Soccer", "Achievements": 7},
    "Michael Phelps": {"Sport": "Swimming", "Achievements": 23},
    "Usain Bolt": {"Sport": "Athletics", "Achievements": 8},
    "Roger Federer": {"Sport": "Tennis", "Achievements": 20},
    "Cristiano Ronaldo": {"Sport": "Soccer", "Achievements": 5}
}

player = players["Lionel Messi"]

if player["Achievements"] > 10:
    print("Player Name:", "Lionel Messi")
    print("Sport:", player["Sport"])
    print("Achievements:", player["Achievements"])
else:
    print("Lionel Messi does not have more than 10 achievements.")
```

```python
for name, details in players.items():
    if details["Sport"] == "Tennis" or details["Achievements"] == 20:
        print("Success:", name, "meets the condition.")
```

```python
for name, details in players.items():
    if details["Achievements"] < 10 and details["Sport"] != "Soccer":
        print("Player Name:", name)
        print("Sport:", details["Sport"])
        print("Achievements:", details["Achievements"])
```

```python
num = int(input("Enter an integer between 1 and 9999: "))

# Check whether the number is even or odd
if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

# Check whether the number is a palindrome
if str(num) == str(num)[::-1]:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")

# Calculate the sum of digits
digit_sum = sum(int(digit) for digit in str(num))

print("Sum of digits:", digit_sum)

# Check divisibility by sum of digits
if num % digit_sum == 0:
    print("The number is divisible by the sum of its digits.")
else:
    print("The number is not divisible by the sum of its digits.")
```

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']

new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print("New squares:", new_squares)
```

```python
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])

    i += 1

print("Animals with 7 letters:", seven_letter_animals)
```
